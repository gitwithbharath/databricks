```python
%sql
create table if not exists workspace.sql_inputs.day1_emp deep clone emp
```

```python
df = spark.table('emp')

df.show()
```

```python
emp_df = spark.table('emp')
mng_df = spark.table('emp')

joined_df = emp_df.join(mng_df, emp_df.manager_id == mng_df.emp_id, 'inner').sort(emp_df.emp_id)

filter_df = joined_df.where(emp_df.salary > mng_df.salary).select(emp_df.emp_name.alias('employee_name'), mng_df.emp_name.alias('manager_name'), emp_df.salary.alias('employee_salary'), mng_df.salary.alias('manager_salary'))

display(filter_df)


```
