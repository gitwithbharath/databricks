```python
%sql

create table entries ( 
name varchar(20),
address varchar(20),
email varchar(20),
floor int,
resources varchar(10));

insert into entries 
values ('A','Bangalore','A@gmail.com',1,'CPU'),('A','Bangalore','A1@gmail.com',1,'CPU'),('A','Bangalore','A2@gmail.com',2,'DESKTOP')
,('B','Bangalore','B@gmail.com',2,'DESKTOP'),('B','Bangalore','B1@gmail.com',2,'DESKTOP'),('B','Bangalore','B2@gmail.com',1,'MONITOR');

select * from entries;
```

```python
%sql

with floor_entries as (select name, floor, count(*) as no_of_floor_entries, rank() over(partition by name order by count(*) desc) as rn
from entries
group by name, floor)
, distinct_resources as (
select distinct name, resources
from entries)
, most_visited_floor as (
select name, floor
from floor_entries
where rn =1
)
select t1.name as name, floor as most_visited_floor, resources
from
(select name, string_agg(resources, ',') as resources
from distinct_resources
group by name) t1
join most_visited_floor t2 on t1.name = t2.name
```
