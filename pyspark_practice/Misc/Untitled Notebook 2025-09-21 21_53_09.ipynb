```python
df = spark.createDataFrame([("Harsha Vardhan"),("Ravi Shankar"),("Bharath kumar Reddy")],["full_name"])
```

```python
df.show()
```

```python
df.write.mode("overwrite").saveAsTable("name_parts")
```

```python
from pyspark.sql.functions import split, size
df = df.withColumn("split", split(df.full_name, " "))
df.withColumn("first_name", df.split[0]).withColumn("last_name", df.split[size(df.split)-1]).show()
```

```python
%sql
select * from name_parts;
```

```python
%sql

select full_name, charindex(" ", full_name)
from name_parts
```

```python
%sql
select full_name, regexp_extract(full_name, r'^(\S+)'), regexp_extract(full_name, r'(\S+)$')
from name_parts
```

```python
df2 = spark.createDataFrame([("Mumbai400001"), ("Delhi110034")], ["city_pincode"])
```

```python
df2.show()
```

```python
df2.select(col("city_pincode")).show()
```

```python
from pyspark.sql.functions import col, length, regexp_extract
df2 = df2.withColumn(
    "city", regexp_extract(col("city_pincode"), r"^([A-Za-z]+)", 1) 
).withColumn(
    "pincode",
    regexp_extract(col("city_pincode"), r"([0-9]+)$", 1)
)
display(df2)
```
